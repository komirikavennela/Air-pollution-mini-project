## Step 2: Survey data (100 residents)
Edit only the `SURVEY` dictionary if your real numbers differ.

```python
import os

OUT = "output"
CH = os.path.join(OUT, "charts")
os.makedirs(CH, exist_ok=True)

# ----------------------------------------------------------------------
# 1. SURVEY DATA  (counts out of TOTAL residents)  -> edit if needed
# ----------------------------------------------------------------------
TOTAL = 100
SURVEY = {
    "aware": 75,
    "not_aware": 25,
    "health_problems": 40,
    "respiratory": 40,
    "cardiovascular": 20,
    "income_agriculture": 50,
    "income_daily_wage": 30,
    "income_other": 20,
    "economic_loss": 20,
}
```

## Step 3: Analysis and output

```python
# ----------------------------------------------------------------------
# 2. ANALYSIS + CONSOLE OUTPUT
# ----------------------------------------------------------------------
def pct(n): return round(n * 100 / TOTAL)

def run_analysis():
    assert SURVEY["aware"] + SURVEY["not_aware"] == TOTAL, "Awareness counts must add up to TOTAL"
    assert SURVEY["income_agriculture"] + SURVEY["income_daily_wage"] + SURVEY["income_other"] == TOTAL
    print("=" * 64)
    print(" MINI PROJECT - AIR POLLUTION SURVEY, GUDIPALLIPADU")
    print(f" Residents surveyed: {TOTAL}")
    print("=" * 64)
    print(f"{'Parameter':<46}{'Count':>6}{'%':>8}")
    print("-" * 64)
    labels = {
        "aware": "Aware of causes and effects",
        "not_aware": "Not aware",
        "health_problems": "Reported health problems",
        "respiratory": "Reported respiratory problems",
        "cardiovascular": "Reported cardiovascular problems",
        "income_agriculture": "Income: agriculture",
        "income_daily_wage": "Income: daily wage labour",
        "income_other": "Income: other sources",
        "economic_loss": "Reported economic loss",
    }
    for k, name in labels.items():
        print(f"{name:<46}{SURVEY[k]:>6}{pct(SURVEY[k]):>7}%")
    print("-" * 64)
    print("Conclusion: awareness is high (%d%%) but %d%% still report health problems." %
          (pct(SURVEY["aware"]), pct(SURVEY["health_problems"])))
    print("=" * 64)

run_analysis()
```

## Step 4: Charts

```python
# ----------------------------------------------------------------------
# 3. CHARTS
# ----------------------------------------------------------------------
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

def make_charts():
    NAVY="#1F3A5F"; TEAL="#2A9D8F"; AMBER="#E9A23B"; RED="#D1495B"; GREY="#C9D1DA"
    plt.rcParams.update({"font.family":"DejaVu Sans","font.size":10})

    def save(fig,name):
        fig.savefig(os.path.join(CH,name),dpi=200,bbox_inches="tight"); plt.close(fig)

    # 1 awareness
    fig,ax=plt.subplots(figsize=(4.2,3.6))
    ax.pie([SURVEY["aware"],SURVEY["not_aware"]],labels=[f"Aware ({SURVEY['aware']})",f"Not aware ({SURVEY['not_aware']})"],colors=[TEAL,GREY],autopct="%1.0f%%",
           startangle=90,wedgeprops=dict(width=0.45,edgecolor="white"),textprops={"fontsize":10})
    ax.set_title("Awareness of causes and effects\n(n = 100) residents",fontsize=11,color=NAVY)
    save(fig,"c1.png")

    # 2 health effects
    fig,ax=plt.subplots(figsize=(5.2,3.4))
    labels=["Respiratory\nproblems","Cardiovascular\nproblems"]; vals=[pct(SURVEY["respiratory"]),pct(SURVEY["cardiovascular"])]
    b=ax.bar(labels,vals,color=[RED,AMBER],width=0.5)
    for r,v in zip(b,vals): ax.text(r.get_x()+r.get_width()/2,v+1.2,f"{v}%",ha="center",fontweight="bold")
    ax.set_ylim(0,55); ax.set_ylabel("% of residents reporting")
    ax.set_title("Health effects reported by residents",fontsize=11,color=NAVY)
    for s in ("top","right"): ax.spines[s].set_visible(False)
    save(fig,"c2.png")

    # 3 income
    fig,ax=plt.subplots(figsize=(5.2,3.4))
    labels=["Agriculture","Daily wage\nlabour","Other sources"]; vals=[pct(SURVEY["income_agriculture"]),pct(SURVEY["income_daily_wage"]),pct(SURVEY["income_other"])]
    b=ax.barh(labels[::-1],vals[::-1],color=[GREY,AMBER,TEAL],height=0.5)
    for r,v in zip(b,vals[::-1]): ax.text(v+1,r.get_y()+r.get_height()/2,f"{v}%",va="center",fontweight="bold")
    ax.set_xlim(0,65); ax.set_xlabel("% of residents")
    ax.set_title("Main source of income",fontsize=11,color=NAVY)
    for s in ("top","right"): ax.spines[s].set_visible(False)
    save(fig,"c3.png")

    # 4 summary
    fig,ax=plt.subplots(figsize=(6.4,3.4))
    labels=["Aware of causes\n& effects","Health problems\nreported","Respiratory\nproblems","Cardiovascular\nproblems","Economic loss\nreported"]
    vals=[pct(SURVEY["aware"]),pct(SURVEY["health_problems"]),pct(SURVEY["respiratory"]),pct(SURVEY["cardiovascular"]),pct(SURVEY["economic_loss"])]; cols=[TEAL,RED,RED,AMBER,NAVY]
    b=ax.bar(labels,vals,color=cols,width=0.55)
    for r,v in zip(b,vals): ax.text(r.get_x()+r.get_width()/2,v+1.5,f"{v}%",ha="center",fontweight="bold")
    ax.set_ylim(0,90); ax.set_ylabel("% of residents (n = 100)")
    ax.set_title("Summary of survey outcomes",fontsize=11,color=NAVY)
    ax.tick_params(axis="x",labelsize=8)
    for s in ("top","right"): ax.spines[s].set_visible(False)
    save(fig,"c4.png")

make_charts()

from IPython.display import Image, display
for n in ("c1", "c2", "c3", "c4"):
    display(Image(os.path.join(CH, n + ".png"), width=420))
```

## Step 5: Build the PDF report

```python
# ----------------------------------------------------------------------
# 4. PDF REPORT
# ----------------------------------------------------------------------
from reportlab.lib.pagesizes import A4
from reportlab.lib.styles import ParagraphStyle
from reportlab.lib.units import cm
from reportlab.lib import colors
from reportlab.lib.enums import TA_CENTER, TA_JUSTIFY
from reportlab.platypus import (BaseDocTemplate, PageTemplate, Frame, Paragraph, Spacer,
                                PageBreak, Table, TableStyle, Image, KeepTogether, ListFlowable, ListItem)

NAVY=colors.HexColor("#1F3A5F"); TEAL=colors.HexColor("#2A9D8F"); LIGHT=colors.HexColor("#EEF3F8")
def build_pdf():
    global W,H,body,h1,h2,cen,cap,cell,cellb,cellh,doc,S
    W,H=A4
    body=ParagraphStyle("body",fontName="Times-Roman",fontSize=11.5,leading=17,alignment=TA_JUSTIFY,spaceAfter=6)
    h1=ParagraphStyle("h1",fontName="Times-Bold",fontSize=16,leading=20,textColor=NAVY,spaceBefore=4,spaceAfter=8)
    h2=ParagraphStyle("h2",fontName="Times-Bold",fontSize=13,leading=17,textColor=NAVY,spaceBefore=8,spaceAfter=4)
    cen=ParagraphStyle("cen",parent=body,alignment=TA_CENTER)
    cap=ParagraphStyle("cap",fontName="Times-Italic",fontSize=10,leading=13,alignment=TA_CENTER,textColor=colors.HexColor("#444444"),spaceAfter=10)
    cell=ParagraphStyle("cell",fontName="Times-Roman",fontSize=10.5,leading=14)
    cellb=ParagraphStyle("cellb",parent=cell,fontName="Times-Bold")
    cellh=ParagraphStyle("cellh",parent=cell,fontName="Times-Bold",textColor=colors.white)

    def P(t,s=body): return Paragraph(t,s)
    def bullets(items):
        return ListFlowable([ListItem(P(i),leftIndent=14) for i in items],bulletType="bullet",start="•",leftIndent=16,bulletFontSize=10)
    def tbl(rows,widths,header=True):
        data=[[Paragraph(str(c),cellh if (header and r==0) else cell) for c in row] for r,row in enumerate(rows)]
        t=Table(data,colWidths=widths,repeatRows=1 if header else 0)
        st=[("GRID",(0,0),(-1,-1),0.5,colors.HexColor("#9AA8B8")),("VALIGN",(0,0),(-1,-1),"MIDDLE"),
            ("TOPPADDING",(0,0),(-1,-1),5),("BOTTOMPADDING",(0,0),(-1,-1),5)]
        if header: st.append(("BACKGROUND",(0,0),(-1,0),NAVY))
        for r in range(1,len(rows)):
            if r%2==0: st.append(("BACKGROUND",(0,r),(-1,r),LIGHT))
        t.setStyle(TableStyle(st)); return t
    def img(path,w):
        from PIL import Image as PI
        iw,ih=PI.open(path).size
        return Image(path,width=w,height=w*ih/iw)

    def footer(c,d):
        if d.page>1:
            c.saveState(); c.setFont("Times-Roman",9); c.setFillColor(colors.HexColor("#555555"))
            c.drawString(2.2*cm,1.2*cm,"Mini Project: Impact of Air Pollution on Health and Environment in Gudipallipadu")
            c.drawRightString(W-2.2*cm,1.2*cm,f"Page {d.page}")
            c.setStrokeColor(colors.HexColor("#B8C3CF")); c.line(2.2*cm,1.6*cm,W-2.2*cm,1.6*cm); c.restoreState()

    doc=BaseDocTemplate(os.path.join("output","Mini_Project_Report_Air_Pollution.pdf"),pagesize=A4,
        leftMargin=2.2*cm,rightMargin=2.2*cm,topMargin=2.2*cm,bottomMargin=2.2*cm,
        title="Mini Project Report - Impact of Air Pollution on Health and Environment in Gudipallipadu",
        author="Komirika Vennela")
    doc.addPageTemplates([PageTemplate(id="p",frames=[Frame(doc.leftMargin,doc.bottomMargin,doc.width,doc.height,id="f")],onPage=footer)])
    S=[]

    # ---------- Cover ----------
    big=ParagraphStyle("big",fontName="Times-Bold",fontSize=20,leading=26,alignment=TA_CENTER,textColor=NAVY)
    S+= [Spacer(1,0.6*cm),P("<b>AP STATE COUNCIL OF HIGHER EDUCATION</b>",cen),
         P("(A STATUTORY BODY OF GOVERNMENT OF ANDHRA PRADESH)",cen),Spacer(1,0.8*cm),
         P("<b>COMMUNITY SERVICE PROJECT</b>",cen),P("<b>MINI PROJECT REPORT</b>",cen),Spacer(1,0.8*cm),
         P("on",cen),Spacer(1,0.3*cm),
         Paragraph("IMPACT OF AIR POLLUTION ON HEALTH AND ENVIRONMENT IN GUDIPALLIPADU",big),
         Spacer(1,0.8*cm),P("By",cen),P("<b>KOMIRIKA VENNELA</b>",cen),P("<b>Reg. No. 24JNA05355</b>",cen),Spacer(1,0.5*cm),
         P("Under the guidance of",cen),P("<b>Ms. Ch. VANAJA, M.Tech, Assistant Professor</b>",cen),
         P("Department of Computer Science and Engineering",cen),Spacer(1,1.0*cm),
         P("<b>SREE VENKATESWARA COLLEGE OF ENGINEERING</b>",cen),
         P("(Accredited with NBA and NAAC 'A' grade, UGC 2(f) Recognized &amp; ISO 9001:2015 Certified)<br/>"
           "(Approved by AICTE, New Delhi and Affiliated to JNTUA, Anantapuramu)<br/>"
           "North Rajupalem, Kodavaluru (V&amp;M), S.P.S.R Nellore (Dt) - 524316",cen),
         Spacer(1,0.6*cm),P("<b>2024 - 2028</b>",cen),PageBreak()]

    # ---------- Project details ----------
    S+=[P("Project Details",h1),
     tbl([["Name of the Student","KOMIRIKA VENNELA"],["Registration Number","24JNA05355"],
          ["Program / Year","B.Tech (Computer Science and Engineering), III Year"],
          ["College","Sree Venkateswara College of Engineering"],
          ["Faculty Guide","Ms. Ch. Vanaja, M.Tech, Assistant Professor"],
          ["Community / Habitation","Gudipallipadu"],
          ["Mini Project Title","Impact of Air Pollution on Health and Environment in Gudipallipadu"],
          ["Mini Project Period","13-07-2026 to 26-07-2026 (2 weeks)"],
          ["Main Project Period","01-06-2026 to 12-07-2026 (6 weeks)"]],[5.2*cm,11.4*cm],header=False),
     Spacer(1,0.6*cm),P("Contents",h1),
     tbl([["1","Introduction"],["2","Objectives"],["3","Methodology"],["4","Questionnaire"],["5","Data Analysis and Results"],
          ["6","Findings"],["7","Discussion"],["8","Recommendations"],["9","Conclusion, Limitations and Future Scope"],["10","References"]],
         [1.5*cm,15.1*cm],header=False),PageBreak()]

    # ---------- 1-3 ----------
    S+=[P("1. Introduction",h1),
     P("Air pollution is the presence of harmful substances in the air at levels that can damage human health, "
       "the environment and the climate. Common sources include vehicle emissions, industrial activity, agricultural "
       "practices, burning of waste and natural events such as dust storms. Exposure is linked to respiratory illness, "
       "cardiovascular disease and wider ecosystem damage."),
     P("During the Community Service Project, awareness campaigns, school workshops and community meetings were conducted "
       "in Gudipallipadu. This mini project builds on that work. It studies how aware residents are of air pollution, "
       "what health effects they report, and what measures they suggest, using a questionnaire-based survey of 100 residents."),
     P("2. Objectives",h1),
     bullets(["To assess the level of awareness about air pollution among residents.",
              "To identify the health effects of air pollution reported by residents.",
              "To suggest measures to mitigate air pollution in the community."]),
     P("3. Methodology",h1),
     tbl([["Step","Activity","Details"],
          ["1","Topic selection and planning","Topic and objectives finalised on 13-07-2026; questionnaire designed."],
          ["2","Survey","Questionnaire-based survey of 100 residents of Gudipallipadu."],
          ["3","Data analysis","Responses tabulated and converted to percentages to find awareness and health effects."],
          ["4","Report preparation","Findings, conclusion and recommendations written and reviewed with the Faculty Guide."]],
         [1.4*cm,4.6*cm,10.6*cm]),Spacer(1,6),
     P("<b>Sample size:</b> 100 residents. <b>Tool:</b> structured questionnaire (Yes/No and short-answer questions). "
       "<b>Analysis:</b> percentage analysis presented through tables and charts."),PageBreak()]

    # ---------- 4 Questionnaire ----------
    S+=[P("4. Questionnaire",h1),
     P("The questionnaire covered four areas. Residents were interviewed politely and without disturbing their households."),
     tbl([["Section","Questions"],
          ["Demographic characteristics","Age, sex, occupation, education level, income level"],
          ["Health status","Do you suffer from any respiratory disease? Have you experienced health problems due to air pollution? "
                           "Have you visited a doctor or hospital for air pollution related issues? (Yes/No)"],
          ["Economic activities","Primary source of income? Do you own agricultural land or livestock? "
                                 "Have you experienced economic losses due to air pollution? (Yes/No)"],
          ["Awareness","Are you aware of the causes and effects of air pollution? Have you taken any measures to reduce exposure? "
                        "Do you think air pollution is a major problem in your area? (Yes/No)"]],[4.6*cm,12*cm]),
     Spacer(1,10),
     P("5. Data Analysis and Results",h1),
     P("The survey responses were analysed and are summarised below. All percentages are out of 100 residents."),
     tbl([["Parameter","Result","Residents (out of 100)"],
          ["Aware of causes and effects of air pollution","75%","75"],
          ["Not aware","25%","25"],
          ["Reported health problems due to air pollution","40%","40"],
          ["Reported respiratory problems","40%","40"],
          ["Reported cardiovascular problems","20%","20"],
          ["Main income: agriculture","50%","50"],
          ["Main income: daily wage labour","30%","30"],
          ["Main income: other sources","20%","20"],
          ["Reported economic losses due to air pollution","20%","20"]],[8.6*cm,3.4*cm,4.6*cm]),
     PageBreak()]

    # ---------- Charts ----------
    S+=[P("5.1 Output: Charts of Survey Results",h1),
     img(os.path.join(CH,"c1.png"),7.6*cm),P("Figure 1: Awareness of causes and effects of air pollution",cap),
     img(os.path.join(CH,"c2.png"),10.5*cm),P("Figure 2: Health effects reported by residents",cap),
     PageBreak(),
     img(os.path.join(CH,"c3.png"),10.5*cm),P("Figure 3: Main source of income of surveyed residents",cap),
     img(os.path.join(CH,"c4.png"),13*cm),P("Figure 4: Summary of survey outcomes",cap),PageBreak()]

    # ---------- 6-7 ----------
    S+=[P("6. Findings",h1),
     bullets(["<b>Awareness:</b> 75% of residents know the causes and effects of air pollution, while 25% do not.",
              "<b>Health effects:</b> 40% report respiratory problems and 20% report cardiovascular problems.",
              "<b>Livelihood:</b> Agriculture (50%) and daily wage labour (30%) are the main sources of income, "
              "so residents spend long hours outdoors.",
              "<b>Economic impact:</b> 20% of residents report economic losses due to air pollution.",
              "<b>Suggested measures:</b> Residents suggested promoting clean energy, improving waste management "
              "practices and increasing green cover."]),
     P("7. Discussion",h1),
     P("Although three out of four residents are aware of air pollution, a large share still report health problems. "
       "This suggests that awareness alone is not enough; practical measures are needed to reduce exposure at the source. "
       "Since many residents work in agriculture and as daily wage labourers, they are exposed to dust, smoke and "
       "burning of waste for long periods. The 25% of residents who are not aware of the causes and effects show that "
       "awareness programmes should continue and reach every household."),
     P("8. Recommendations",h1),
     tbl([["Recommendation","Action"],
          ["Promote clean energy","Encourage solar, wind and biogas to reduce dependence on fossil fuels."],
          ["Improve waste management","Segregation, recycling and composting to reduce waste burning."],
          ["Increase green cover","Plant trees and develop green spaces to absorb pollutants."],
          ["Raise awareness","Continue campaigns, school programmes and community meetings."],
          ["Monitor air quality","Set up air quality monitoring to track pollutant levels and guide decisions."]],[5*cm,11.6*cm]),
     PageBreak()]

    # ---------- 9-10 ----------
    S+=[P("9. Conclusion, Limitations and Future Scope",h1),
     P("<b>Conclusion.</b> The mini project shows that air pollution has a significant impact on health and environment in "
       "Gudipallipadu. Awareness is fairly high (75%), but 40% of residents still report health problems and 20% report "
       "economic losses. Promoting clean energy, improving waste management and increasing green cover can help "
       "reduce air pollution, along with continued awareness programmes."),
     P("<b>Limitations.</b> The study was limited to Gudipallipadu and a sample of 100 residents, and the results are based on "
       "self-reported responses. It may not be representative of other areas."),
     P("<b>Future scope.</b> Further studies can assess the impact of air pollution on specific health outcomes, evaluate "
       "the effectiveness of awareness programmes, and develop strategies to promote clean energy and better waste management."),
     P("10. References",h1),
     bullets(["World Health Organization (WHO): Air pollution and health fact sheets.",
              "Central Pollution Control Board (CPCB), Government of India: National Air Quality Index.",
              "AP State Council of Higher Education: Guidelines for the OJT Internship / Community Service Project.",
              "Primary data: questionnaire survey of 100 residents of Gudipallipadu, July 2026."]),
     Spacer(1,1.5*cm),
     Table([[P("Signature of the Student",cen),P("Signature of the Faculty Guide",cen)]],colWidths=[8.3*cm,8.3*cm])]

    doc.build(S)
    print("PDF created: output/Mini_Project_Report_Air_Pollution.pdf")

build_pdf()
```
